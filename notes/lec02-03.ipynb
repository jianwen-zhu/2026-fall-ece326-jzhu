## Recursion
-----

Problem solving technique that mimics mathematical induction.

- Define base case
- Recurse
- Avoid infinite recursion!


````python
def count_down( n ) :
    if n <= 0 :
        print( 'Blastoff!' )
    else :
        print( n )
        count_down( n-1 )
````

````python
count_down( 10 )
````

### Compute Factorial
  - Base Case: 0! = 1
  - Recursion: n! = n ( n - 1 )!


````python
def factorial( n ) :
    if n == 0 :
        return 1            # basecase
    else :
        recurse = factorial( n - 1 )
        result = n * recurse
        return result
````

````python
factorial( 10 )
````

### Adding Debugging Info


````python
def factorial( n ) :
    space = ' ' * (4*n)   # amount of identation with repeat operator
    print( space + 'calling ' + n.__str__() )
    if n == 0 :
        print( space + 'returning 1' )
        return 1          
    else :
        recurse = factorial( n - 1 )
        result = n * recurse
        print( space + 'returning' + result.__str__() )
        return result
````

````python
factorial( 10 )
````

### Runtime Type Checking

What would happen if we have factorial( 1.5 )	
  - Add type checking


````python
factorial( 1.5 )
````

````python
def factorial( n ) :
    if not isinstance(n,int) :   #runtime type checking, not in C
        return None
    if n == 0 :
        return 1          
    else :
        recurse = factorial( n - 1 )
        result = n * recurse
        return result
````

````python
factorial( 1.5 )
````

````python
factorial( 10 )
````

Let's also do Fibonacci


````python
def fibonacci( n ) :
    if n < 2 :
        return n
    else :
        recurse1 = fibonacci( n - 1 )
        recurse2 = fibonacci( n - 2 )
        return recurse1 + recurse2
````

### Executing Module as a Script


````python
%%file fib.py
import sys

def fibonacci( n ) :
    if n < 2 :
        return n
    else :
        recurse1 = fibonacci( n - 1 )
        recurse2 = fibonacci( n - 2 )
        return recurse1 + recurse2

if __name__ == "__main__":
    print( fibonacci(int(sys.argv[1])) )
````
